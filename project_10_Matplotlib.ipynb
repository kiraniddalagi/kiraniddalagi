```python
'''1. Import Required Libraries

In Python, we usually import matplotlib.pyplot to draw charts and pandas to work with data tables. Notes pyplot is the plotting interface used for most Matplotlib charts. pandas helps store and inspect tabular data. In Jupyter Notebook, plots appear inline when the notebook is configured correctly. '''
import pandas as pd
import matplotlib.pyplot as plt

print('Matplotlib is ready!')
```

```python
'''2. Load and Inspect Sample Data '''
sales_data = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'sales': [120, 150, 170, 160, 210, 250]
})

print(sales_data)
print('\nShape:', sales_data.shape)
print('\nColumns:', sales_data.columns.tolist())
print('\nHead:')
print(sales_data.head())
print('\nDescription:')
print(sales_data.describe())
```

```python
''' 3. Create a Simple Line Plot'''

plt.figure(figsize=(8, 5))
plt.plot(sales_data['month'], sales_data['sales'], color='blue', marker='o', linewidth=2)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
'''4. Create a Bar Chart '''

categories = ['A', 'B', 'C', 'D']
values = [25, 40, 30, 60]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color=['#4c72b0', '#55a868', '#c44e52', '#8172b2'])
plt.title('Product Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
for i, v in enumerate(values):
    plt.text(i, v + 2, str(v), ha='center')
plt.tight_layout()
plt.show()
```

```python
''' 5. Create a Scatter Plot'''

x = [1, 2, 3, 4, 5, 6, 7]
y = [2, 4, 5, 8, 10, 11, 15]

plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='darkgreen', s=80, alpha=0.8)
plt.title('Study Hours vs Grades')
plt.xlabel('Hours Studied')
plt.ylabel('Score')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
''' 6. Create a Histogram'''

scores = [55, 60, 65, 66, 70, 72, 74, 75, 80, 82, 85, 90, 92, 95]

plt.figure(figsize=(8, 5))
plt.hist(scores, bins=5, color='purple', edgecolor='black')
plt.title('Distribution of Exam Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
'''7.. Create a Pie Chart '''
labels = ['A', 'B', 'C', 'D']
sizes = [30, 15, 30, 40]

plt.figure(figsize=(7, 7))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['#66c2a5', '#fc8d62', '#8da0cb', '#e78ac3'])
plt.title('Market Share Distribution')
plt.tight_layout()
plt.show()
```

```python
'''8. Customize Plot Styling '''
plt.figure(figsize=(9, 5))
plt.plot([1, 2, 3, 4, 5], [10, 15, 13, 18, 22],
         color='red', linestyle='--', linewidth=2.5, marker='s', markersize=8)
plt.title('Styled Line Chart')
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
''' 9. Add Labels, Titles, and Legends'''
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
product_a = [20, 25, 23, 30, 35]
product_b = [15, 18, 22, 26, 31]

plt.figure(figsize=(8, 5))
plt.plot(months, product_a, marker='o', label='Product A')
plt.plot(months, product_b, marker='o', label='Product B', linestyle='--')
plt.title('Monthly Comparison')
plt.xlabel('Month')
plt.ylabel('Units Sold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
'''10. Save Plot and Final Notes '''

plt.figure(figsize=(8, 5))
plt.plot(sales_data['month'], sales_data['sales'], color='green', marker='o')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.tight_layout()
plt.savefig('monthly_sales.png', dpi=300)
plt.show()
print('Chart saved as monthly_sales.png')
```
